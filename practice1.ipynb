```python
import pandas as pd

TARGET_PATH = 'data/data.csv'
```

```python
df = pd.read_csv(TARGET_PATH, encoding='euc-kr')
```

```python
# 수출물동량 구해보기
```

```python
total = df.groupby('청코드')['전체물동량'].sum()
```

```python
# 수출인것들의 전체물동량 합 구하면 되는거아님?

export = df[df['수출입구분명'] == ('수출')]
export
```

```python
export_total = export.groupby('청코드')['전체물동량'].sum()
```

```python
(export_total / total * 100).round(2)
```

```python
import_con = df[df['수출입구분명'] == ('수입')]
```

```python
import_total = import_con.groupby('청코드')['전체물동량'].sum()
```

```python
(import_total / total * 100).round(2)
```
